```python
import pandas as pd
df = pd.read_csv("../data/raw/Fraud_Data.csv")

print(df.shape); print(df.dtypes)
print(df.isna().sum())
print("duplicates:", df.duplicated().sum())
print(df["class"].value_counts(normalize=True))
print("unique IPs:", df["ip_address"].nunique())

for c in ["signup_time", "purchase_time"]:
    df[c] = pd.to_datetime(df[c])
print(df.purchase_time.min(), df.purchase_time.max())
print("purchase before signup:", (df.purchase_time < df.signup_time).sum())
print("max users on one device:", df.groupby("device_id").user_id.nunique().max())
print(df[["age", "purchase_value"]].describe())
```
